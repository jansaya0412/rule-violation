```python
import pandas as pd
import os
import json

from ruler.features.make_splits import read_split
```

```python
for split in range(10):
    with open(f'../data/interim/splits/nonbinary/{0}/test_n_rules_out.json', 'r', encoding='utf8') as f:
        test = json.load(f)
    df = pd.DataFrame(((i['applied_rule_n'], i['applied_rule_text'], i['community']['actor_id']) for i in test), columns = ['rule_n', 'rule_text', 'community'])
    print(f'split {split} has {len(df.drop_duplicates())} unique rules in test n rules out: {" ".join(df.drop_duplicates().apply(lambda row: f"{row.community} {row.rule_n}", axis=1))}')
```

```python
for split in range(10):
    _, _, test, _ = read_split(split_n = split, base_folder='../data/interim/splits/')
    df = pd.DataFrame(((i['applied_rule_n'], i['applied_rule_text'], i['community']['actor_id']) for i in test), columns = ['rule_n', 'rule_text', 'community'])
    print(f'split {split} has {len(df.drop_duplicates())} unique rules in test n rules out: {" ".join(df.drop_duplicates().apply(lambda row: f"{row.community} {row.rule_n}", axis=1))}')
```
